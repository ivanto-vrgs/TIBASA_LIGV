```python
import random
import math

def peso_triangular_inversa():
    u = random.random() # Genera numero entre U(0,1)
    # Aplicamos la función calculada con F^-1=u
    if u <= 0.5:
        return 190 + math.sqrt(800 * u)
    else:
        return 230 - math.sqrt(800 * (1 - u))

# Por simulación
iteraciones = 100000
excesos = 0

for _ in range(iteraciones):
    # Sumamos 5 tinas generadas manualmente
    peso_viaje = sum(peso_triangular_inversa() for _ in range(5))
    if peso_viaje > 1000:
        excesos += 1

probabilidad = excesos / iteraciones
print(f"Probabilidad por Transformada Inversa: {100*probabilidad:.4f}%")
```
